```python
import tensorflow as tf
from tensorflow import keras
from keras import layers
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score
from sklearn.datasets import make_classification, make_regression

import warnings
warnings.filterwarnings('ignore')
```

```python
# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)
```

```python
# 1. Generate Synthetic dataset
X_clf, y_clf = make_classification(n_samples=1000, n_features=10, n_informative=5, n_classes=2 , random_state=42)
```

```python
# 2. Split the Data
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(X_clf , y_clf,test_size=0.2, random_state=42)
```

```python
# Standardized Features
scaler_clf = StandardScaler()
```

```python
X_clf_train_scaled = scaler_clf.fit_transform(X_clf_train)
X_clf_test_scaled = scaler_clf.transform(X_clf_test)
```

```python
# 4. Build The Model
model_clf = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_clf_train_scaled.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
# 5. Compile the Model
model_clf.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# 6. Train the Model
history_clf = model_clf.fit(X_clf_train_scaled, y_clf_train, epochs=50 , batch_size=32, verbose=0, validation_split=0.1)
```

```python
# 7. Evaluate the Model
y_clf_pred_proba = model_clf.predict(X_clf_train_scaled).flatten()
```

```python
y_clf_pred = (y_clf_pred_proba > 0.5 ).astype(int)
```

```python
accuracy_clf = accuracy_score(y_clf_test, y_clf_pred)
```
