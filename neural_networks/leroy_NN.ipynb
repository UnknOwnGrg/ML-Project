# Regression Problems

Exercise 1: Synthetic Regression Problem
Task: Create a simple neural network to predict a continuous target variable from synthetic features.

1. Generate a synthetic dataset with n_samples=1000, n_features=10, n_informative=5, random_state=42 using sklearn.datasets.make_regression.
2. Split the data into training and testing sets.
3. Standardize the features using StandardScaler.
4. Build a simple Keras Sequential model with a few Dense layers (e.g., 2-3 layers with ReLU activation) and a final Dense layer with no activation (for linear output).
5. Compile the model using the adam optimizer and mean_squared_error as the loss function.
6. Train the model on the training data.
7. Evaluate the model's performance on the test set using Mean Squared Error (MSE) and R-squared (r2_score).

```python
# 1. Generate synthetic dataset
X_reg, y_reg = make_regression(n_samples=1000, n_features=5, random_state=42)
```

```python
# 2. Split Data
X_reg_train, X_reg_test, y_reg_train, y_reg_test= train_test_split(X_reg, y_reg,test_size=0.2, random_state=42)
```

```python
# 3. Standardize features
scaler_reg = StandardScaler()
```

```python
X_reg_train_scaled = scaler_reg.fit_transform(X_reg_train)
X_reg_test_scaled = scaler_reg.transform(X_reg_test)
```

```python
# 4. Build the Models
model_reg = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_reg_train_scaled.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

```python
# 5. Compile the Model
model_reg.compile(optimizer='adam', loss="mean_squared_error")
```

```python
# 6. Train the Model
history_reg = model_reg.fit(X_reg_train_scaled,y_reg_train, epochs=50, batch_size=32, verbose=0, validation_split=0.1)
```

```python
# 7. Evaluate the Model
y_reg_pred = model_reg.predict(X_reg_test_scaled).flatten()
```

```python
mse_reg = mean_squared_error(y_reg_test, y_reg_pred)
r2_reg = r2_score(y_reg_test, y_reg_pred)

print(f"Exercise 1 - Synthetic Regression:")
print(f"  Mean Squared Error: {mse_reg:.4f}")
print(f"  R-squared: {r2_reg:.4f}")
```
