# California Housing Price Prediction

Task: Predict California housing prices using a neural network. This dataset is more complex than the synthetic one and requires proper preprocessing.

1. Load the California housing dataset using sklearn.datasets.fetch_california_housing().
2. Split the data into training and testing sets.
3. Standardize the features using StandardScaler.
4. Build a slightly more complex Keras Sequential model than in Exercise 1. Experiment with more layers or neurons if needed.
5. Compile the model using the adam optimizer and mean_squared_error as the loss function.
6. Train the model on the training data.
7. Evaluate the model's performance on the test set using Mean Squared Error (MSE) and R-squared (r2_score).

```python
import tensorflow as tf
from tensorflow import keras
from keras import layers
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.metrics import mean_squared_error, r2_score, accuracy_score

from sklearn.datasets import fetch_california_housing

import warnings

warnings.filterwarnings('ignore')
```

```python
california_housing = fetch_california_housing(as_frame=True)
```

```python
X_cali, y_cali  = california_housing.data, california_housing.target
```

```python
X_cali.head()
```

```python
#Split the Data
X_cali_train, X_cali_test, y_cali_train, y_cali_test = train_test_split(X_cali, y_cali, test_size=0.2, random_state=42)
```

```python
# Standardize features
scaler_cali = StandardScaler()
```

```python
X_cali_train_scaled = scaler_cali.fit_transform(X_cali_train)
X_cali_test_scaled = scaler_cali.transform(X_cali_test)
```

```python
# Build the Model
model_cali = keras.Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_cali_train_scaled.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

```python
# 5. Compile the Model
model_cali.compile(optimizer='adam', loss='mean_squared_error')
```

```python
# 6. Train the Model
history_call = model_cali.fit(X_cali_train, y_cali_train, epochs=100, batch_size=64, verbose=0, validation_split=0.1)
```

```python
# 7. Evaluate the Model
y_cali_pred= model_cali.predict(X_cali_test_scaled).flatten()
mse_cali  = mean_squared_error(y_cali_test, y_cali_pred)
r2_cali = r2_score(y_cali_test, y_cali_pred)
```

```python
print(f"Exercise 2 - California Housing Regression:")
print(f"  Mean Squared Error: {mse_cali:.4f}")
print(f"  R-squared: {r2_cali:.4f}")
```
