```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras import  layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import fetch_california_housing

import  warnings
warnings.filterwarnings('ignore')
```

```python
california_housing = fetch_california_housing(as_frame=True)
```

```python
X_cali , y_cali = california_housing.data , california_housing.target
```

```python
X_cali.head()
```

```python
# Split the Data
X_cali_trian , X_cali_test, y_cali_train, y_cali_test = train_test_split(X_cali, y_cali, test_size=0.2, random_state=42)
```

```python
scaler_cali = StandardScaler()
```

```python
X_cali_train_scaled = scaler_cali.fit_transform(X_cali_trian)
X_cali_test_scaled = scaler_cali.transform(X_cali_test)
```

```python
# Build the Model

model_cali = keras.Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_cali_train_scaled.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

```python
# Compile the Model
model_cali.compile(optimizer='adam', loss='mean_squared_error')
```

```python
history_cali = model_cali.fit(X_cali_trian, y_cali_train, epochs=100, batch_size = 64, verbose=0, validation_split=0.1)
```

```python
# Evaluate the Model
y_cali_pred = model_cali.predict(X_cali_test_scaled).flatten()
mse_cali = mean_squared_error(y_cali_test, y_cali_pred)
r2_cali = r2_score(y_cali_test, y_cali_pred)
```

```python
print(f"Exercise 2 - California Housing Regression:")
print(f"  Mean Squared Error: {mse_cali:.4f}")
print(f"  R-squared: {r2_cali:.4f}")
```
