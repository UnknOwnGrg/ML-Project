```python
import tensorflow
from tensorflow import keras
from keras import layers
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

import warnings
warnings.filterwarnings('ignore')
```

```python
iris = load_iris()
# Simply the iris.data will provide columns and target will give the species
X_iris , y_iris = iris.data , iris.target
```

```python
# Split the data
X_iris_train, X_iris_test , y_iris_train, y_iris_test = train_test_split(X_iris, y_iris, test_size=0.2, random_state=42, stratify=y_iris)
```

```python
# Standardize Features
scaler_iris = StandardScaler()
X_iris_train_scaled = scaler_iris.fit_transform(X_iris_train)
X_iris_test_scaled = scaler_iris.fit_transform(X_iris_test)
```

```python
# 4.One-hot Encode the target variable
encoder_iris = OneHotEncoder(sparse_output=False)
y_iris_train_encoded = encoder_iris.fit_transform(y_iris_train.reshape(-1, 1))
y_iris_test_encoded = encoder_iris.transform(y_iris.reshape(-1, 1))
```

```python
# 5. Build The Model
num_classes = len(np.unique(y_iris))
model_iris = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_iris_train_scaled.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
    ])
```

```python
# Compile The Model
model_iris.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# Train the Model
history_iris = model_iris.fit(X_iris_train_scaled, y_iris_train_encoded, epochs=100, batch_size=16 , verbose=1
                              , validation_split=0.1)
```

```python
y_iris_pred_proba = model_iris.predict(X_iris_test_scaled)
y_iris_pred = np.argmax(y_iris_pred_proba ,axis=1)
accuracy_iris = accuracy_score(y_iris_test, y_iris_pred)
```

```python
print(f"Accuracy: {accuracy_iris:.4f}")
```
